```python
# Remove the hyphen from SKU_ID
vendors['SKU_ID'] = vendors['SKU_ID'].str.replace('-', '', regex=False)

# Now re-run your validation check
pattern = r'^SKU\d{5}$'
sku_valid = vendors['SKU_ID'].astype(str).str.match(pattern)

print("SKU_ID valid now:", sku_valid.all())
```

```python
# 1. Strip whitespace and make everything uppercase
vendors['Store_ID'] = vendors['Store_ID'].astype(str).str.strip().str.upper()

# 2. Remove any existing hyphens so we can re-insert them correctly
vendors['Store_ID'] = vendors['Store_ID'].str.replace('-', '', regex=False)
# 3. Fix single digits: If it's 3 letters + 1 digit (e.g., HYD1 -> HYD01)
vendors['Store_ID'] = vendors['Store_ID'].str.replace(
    r'^([A-Z]{3})(\d{1})$', 
    r'\g<1>0\g<2>',  # Use \g<1> to keep group 1 separate from the zero
    regex=True
)

# 4. Insert the hyphen: Turn 3 letters + 2 digits into AAA-00 (e.g., HYD01 -> HYD-01)
vendors['Store_ID'] = vendors['Store_ID'].str.replace(
    r'^([A-Z]{3})(\d{2})$', 
    r'\g<1>-\g<2>', 
    regex=True
)
```

Sales Dataset

```python
sales=pd.read_excel(r'C:\Users\hites\OneDrive\Desktop\Retail Chain Profit Leakage & Demand Optimization\Excel\Raw\sales_raw.xlsx')
sales
```

```python
sales.Order_ID.duplicated()
```

```python
sales = sales.drop_duplicates(subset='Order_ID')
```

```python
sales.isnull().sum()
```

```python
# Create a filtered dataframe for analysis
sales = sales[sales['Order_Status'] != 'Cancelled'].copy()

```

```python
# Get valid promo ids from promotions table
valid_promo_ids = promotions['Promo_ID'].dropna().unique()

# Replace invalid Promo_IDs in sales with NaN
sales.loc[
    ~sales['Promo_ID'].isin(valid_promo_ids),
    'Promo_ID'
] = pd.NA

# View result
sales['Promo_ID'].unique()
```

```python
sales
```

```python
sales['Order_Date'] = pd.to_datetime(
    sales['Order_Date'],
    dayfirst=True,
    errors='coerce'
)
```

```python
sales.dtypes
```

```python


# 1. Clean Store_ID: Ensure uppercase and correct dash placement (e.g., hyd03 -> HYD-03)
# We use regex to insert a dash if it's missing between letters and numbers
sales['Store_ID'] = sales['Store_ID'].str.upper().str.replace(r'([A-Z]{3})(\d{2})', r'\1-\2', regex=True)

# 2. Clean Customer_ID: Ensure 'CUST' prefix and 7-digit padding (e.g., CUST981 -> CUST000981)
# This extracts the digits and pads them to a total length of 10 characters including 'CUST'
sales['Customer_ID'] = sales['Customer_ID'].str.upper().str.extract(r'(\d+)').iloc[:,0]
sales['Customer_ID'] = 'CUST' + sales['Customer_ID'].str.zfill(6)

# 3. Clean SKU_ID: Ensure 'SKU' prefix and 5-digit padding (e.g., SKU565 -> SKU00565)
# Similar to Customer_ID, we extract the numeric part and re-standardize
sales['SKU_ID'] = sales['SKU_ID'].str.upper().str.extract(r'(\d+)').iloc[:,0]
sales['SKU_ID'] = 'SKU' + sales['SKU_ID'].str.zfill(5)

# List of columns to clean
cols_to_fix = [ 'Customer_ID', 'SKU_ID']

# Loop through and remove hyphens
for col in cols_to_fix:
    sales[col] = sales[col].str.replace('-', '', regex=False)

# Display results to verify
print(sales[cols_to_fix].head())
```
